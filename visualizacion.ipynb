```python
df = pd.read_csv("DDoSdata.csv", low_memory=False)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x='attack', data=df)
plt.title("Distribución: tráfico normal vs ataque")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x='category', data=df)
plt.xticks(rotation=45)
plt.title("Tipos de ataques")
plt.show()
```

```python
sns.countplot(x='proto', data=df)
plt.title("Protocolos en el tráfico")
plt.show()
```

```python
sns.boxplot(x='attack', y='bytes', data=df)
plt.title("Bytes por tipo de tráfico")
plt.show()
```

```python
sns.histplot(df['pkts'], kde=True)
plt.title("Distribución de paquetes")
plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(numeric_only=True), cmap='coolwarm')
plt.title("Matriz de correlación")
plt.show()
```
